```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset= pd.read_csv('data/titanic.csv')
dataset.head()
```

```python
dataset['Fare'].mean()
mn= np.mean(dataset['Fare'])

#dataset['Fare']= dataset['Fare'].fillna(dataset['Fare'].mean())
md =np.median(dataset['Fare'])
mod= dataset['Fare'].mode()[0]
```

```python
a=sns.histplot(x='Fare', data=dataset, bins= [i for i in range(0, 81 , 10)])
a.axvline(mn, color='red', linestyle='-.', label='mean'),
a.axvline(md, color='blue', linestyle='-.', label='median')
a.axvline(mod, color='green', linestyle='-.', label='mode')
plt.legend()
#plt.plot([mn for i in range (0,300)], [i for i in range (0,300)], c='red')
```
